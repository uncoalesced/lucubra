# Harmonic Oscillator

Potential energy varies quadratically with displacement from equilibrium.


```python
def potential_energy(stiffness: float, displacement: float) -> float:
    return 0.5 * stiffness * displacement * displacement


print(potential_energy(10.0, 0.5))
```

```python
def spring_force(stiffness: float, displacement: float) -> float:
    return -stiffness * displacement


abs(spring_force(5.0, 0.5))
```
